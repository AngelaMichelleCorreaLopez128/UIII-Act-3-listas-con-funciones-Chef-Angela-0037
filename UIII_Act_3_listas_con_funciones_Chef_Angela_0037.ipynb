#### Lista con funciones

```python
# 1. Base de datos (lista de diccionarios)
chefs = []

# 2. Definición de Funciones
def agregar_chef():
    id_c = input("ID: ")
    nom = input("Nombre: ")
    esp = input("Especialidad: ")
    chefs.append({"id": id_c, "nombre": nom, "especialidad": esp})
    print("-> Guardado.")

def listar_chefs():
    if not chefs:
        print("Lista vacía.")
    for c in chefs:
        print(f"[{c['id']}] {c['nombre']} - {c['especialidad']}")

def buscar_chef():
    id_b = input("ID a buscar: ")
    for c in chefs:
        if c['id'] == id_b:
            print(f"Encontrado: {c['nombre']}")
            return c
    print("No existe.")
    return None

def modificar_chef():
    c = buscar_chef()
    if c:
        c['nombre'] = input(f"Nuevo nombre ({c['nombre']}): ") or c['nombre']
        print("Modificado.")

def borrar_chef():
    id_b = input("ID a borrar: ")
    global chefs
    chefs = [c for c in chefs if c['id'] != id_b]
    print("Proceso terminado.")

# 3. EL MOTOR (Menú Principal)
# Sin esto, el código "no jala" porque nadie llama a las funciones.
def menu():
    while True:
        print("\n--- SISTEMA DE CHEF ---")
        print("1. Agregar | 2. Listar | 3. Buscar | 4. Modificar | 5. Borrar | 6. Salir")
        opcion = input("Selecciona una opción: ")

        if opcion == "1": agregar_chef()
        elif opcion == "2": listar_chefs()
        elif opcion == "3": buscar_chef()
        elif opcion == "4": modificar_chef()
        elif opcion == "5": borrar_chef()
        elif opcion == "6": break
        else: print("Opción no válida.")

# 4. Ejecución
if __name__ == "__main__":
    menu()
print("Programa Realizado por Angela Michelle Correa Lopez NL = 13")
```
